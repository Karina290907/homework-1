```python
# Функція, яка формує текст заяви
def get_application_text(receiver_name: str, course_name: str, date: str, name: str) -> str:
    return f"""
    {receiver_name}

                ЗАЯВА

Прошу зарахувати курс '{course_name}' на 10 навчальних кредитів.

Дата: {date}.         {name}
"""

# Основна частина програми — отримання даних від користувача
receiver = input("Кому адресовано заяву (наприклад: Декану факультету): ")
course = input("Введіть назву курсу: ")
date = input("Введіть дату подання заяви (дд-мм-рррр): ")
name = input("Введіть ПІБ заявника: ")

# Вивід готового документу
print(get_application_text(receiver, course, date, name))
```
